### Examples of generating `EccHybIMR22` using PyCBC

```python
from pycbc.waveform import get_td_waveform
import matplotlib.pyplot as plt
import numpy as np
import time
```

```python
pip show EccHybIMR22
```

### Example 1: $M = 40M_\odot;\ q = 1;\ e_0=0.02;\ l_0=0;\ f_{\rm 0}=20Hz$

```python
M = 40
q = 1
m1 = M*q/(1+q); m2 = M-m1
e = 0.2
l =  0
f0 = 20
inc = 0
dist = 1
dt = 1./4096

t1 = time.time()

hp, hc = get_td_waveform(
    approximant  = "EccHybIMR22",
    mass1        = m1,
    mass2        = m2,
    delta_t      = dt,
    f_lower      = f0,
    distance     = dist,
    inclination  = inc,
    eccentricity = e,
    mean_anomaly = l
)

t2 = time.time()
print('waveform generated in ',t2-t1,' seconds')
```

```python
plt.figure(figsize=(12,3))
plt.plot(hp.sample_times,abs(hp-1j*hc), label='Amplitude',linewidth=1, color='red')
plt.plot(hp.sample_times, hp, color='dodgerblue',label=r'$h_+$',linewidth=1)
plt.plot(hp.sample_times, hc, color='saddlebrown',label=r'$h_\times$',linewidth=1)
plt.xlabel('Time'), plt.ylabel(r'$Re[h_{22}]$')
plt.legend(frameon=False)
```
