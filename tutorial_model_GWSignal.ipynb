### Examples of generating `EcchybIMR_22` using GWSignal

```python
from lalsimulation.gwsignal.models import gwsignal_get_waveform_generator
import astropy.units as u
import time
import matplotlib.pyplot as plt
import numpy as np
```

```python
pip show EcchybIMR_22
```

```python
gen = gwsignal_get_waveform_generator("EcchybIMR_22")
```

### Example 1: $M = 40M_\odot;\ q = 1;\ e_0=0.02;\ l_0=0;\ f_{\rm 0}=20Hz$

```python
M = 40
q = 1
mass1 = M*q/(1+q); mass2 = M-mass1
e = 0.02
l = 0
f0 = 20
inc = 0
dist = 10
dt = 1./4096

t1 = time.time()

params = dict(
    mass1       = mass1 * u.Msun,
    mass2       = mass2 * u.Msun,
    eccentricity= e * u.dimensionless_unscaled,
    meanPerAno  = l * u.rad,
    f22_start   = f0 * u.Hz,
    deltaT      = (dt) * u.s,
    distance    = dist * u.Mpc,
    inclination = inc * u.rad
)

hlm = gen.generate_td_modes(**params)  # generates mode
hp, hc = gen.generate_td_waveform(**params) # generates polarizations

t2 = time.time()
print('Generation time: ',t2-t1,' seconds')
```

```python
plt.figure(figsize=(12,3))
plt.plot(hlm[(2,2)].times.value, hlm[(2,2)].value.real, label=r"$Re(h_{\rm 22})$")
plt.plot(hlm[(2,2)].times.value, hlm[(2,2)].value.imag, label=r"$Im(h_{\rm 22})$")
plt.plot(hlm[(2,2)].times.value, abs(hlm[(2,2)].value), label='Amplitude',linewidth=1, color='black')
plt.xlabel("Time (s)", fontsize=10)
plt.grid(None)
plt.legend(frameon=False, fontsize=10)
plt.title('(2, 2) Mode',fontsize=15)
plt.show()

plt.figure(figsize=(12,3))
plt.plot(hp.times.value, hp.value, label=r"$h_+$")
plt.plot(hc.times.value, hc.value, label=r"$h_\times$")
plt.plot(hp.times.value, abs(hp.value-1j*hc.value), label='Amplitude',linewidth=1, color='black')
plt.xlabel("Time (s)")
plt.grid(None)
plt.title('Polarizations',fontsize=15)
plt.legend(frameon=False, fontsize=10)
plt.xlabel("Time (s)", fontsize=10)
plt.show()
```
